## TPC9 - PLNEB

```python
import spacy

nlp = spacy.load("pt_core_news_sm")


import sys
print(sys.executable)
```

```python
# Abrir e ler livros

with open("Harry Potter e A Pedra Filosofal.txt", "r", encoding="utf-8") as f:
    livro1 = f.read()

with open("Harry_Potter_Camara_Secreta-br.txt", "r", encoding="utf-8") as f:
    livro2 = f.read()
```

```python
# Processar com spacy

docs = []

for texto in [livro1, livro2]:
    doc = nlp(texto)
    docs.append(doc)
```

```python
# Criar listas de tokens

sentences = []

for doc in docs:
    for sent in doc.sents:
        tokens = [token.text.lower() for token in sent if token.is_alpha]
        if tokens:
            sentences.append(tokens)
```

```python
# Treinar Word2Vec

from gensim.models import Word2Vec

model = Word2Vec(
    sentences,
    vector_size=100,
    window=5,
    min_count=2,
    sg=1
)

model_wv = model.wv
```

```python
# Similarity
print(model_wv.similarity("harry", "hermione"))
print(model_wv.similarity("harry", "draco"))
print(model_wv.similarity("harry", "malfoy"))
print(model_wv.similarity("harry", "gina"))
```

```python
# Doesnt_match
print(model_wv.doesnt_match(["harry", "rony", "hermione", "dumbledore"]))
print(model_wv.doesnt_match(["gryffindor", "slytherin", "hufflepuff", "harry"]))
print(model_wv.doesnt_match(["harry", "rony", "hermione", "hagrid"]))
```

```python
# Most_similar
print(model_wv.most_similar("harry"))
print(model_wv.most_similar("rony"))
print(model_wv.most_similar("hermione"))
print(model_wv.most_similar("hogwarts"))
```

```python
# Guardar o modelo Word2Vec em formato word2vec text
model_wv.save_word2vec_format('model_harry.txt', binary=False)
```

```python
# Converter para formato TensorFlow (tensor + metadata)
#!python3 -m gensim.scripts.word2vec2tensor -i model_harry.txt -o model_harry
!/usr/local/bin/python3.10 -m gensim.scripts.word2vec2tensor -i model_harry.txt -o model_harry
```
